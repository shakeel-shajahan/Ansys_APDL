# 10. Linear Regression: Main Idea

Suppose we have measured points:

\[
(x_1,y_1), (x_2,y_2), \ldots, (x_m,y_m)
\]

We want to find a straight line:

\[
g(x) = ax + b
\]

The line will not pass exactly through all points.  
So we measure the error.

For one point:

\[
\text{error}_i = ax_i + b - y_i
\]

For all points, we use squared error:

\[
f(a,b) = \sum_{i=1}^{m} (ax_i + b - y_i)^2
\]

Why square?

1. Negative and positive errors do not cancel each other.
2. Large mistakes become more important.
3. The formula becomes smooth and easy to minimize.

# 11. Exercise Task 1(a): Derivation of the Linear System

We want to minimize:

\[
f(a,b) = \sum_{i=1}^{m} (ax_i + b - y_i)^2
\]

Let:

\[
r_i = ax_i + b - y_i
\]

Then:

\[
f(a,b) = \sum_{i=1}^{m} r_i^2
\]

To find the best line, we differentiate with respect to \(a\) and \(b\).

---

## Derivative with respect to \(a\)

\[
\frac{\partial f}{\partial a}
=
\sum_{i=1}^{m} 2r_i \frac{\partial r_i}{\partial a}
\]

Since:

\[
r_i = ax_i + b - y_i
\]

we get:

\[
\frac{\partial r_i}{\partial a} = x_i
\]

So:

\[
\frac{\partial f}{\partial a}
=
2\sum_{i=1}^{m}(ax_i+b-y_i)x_i
\]

At the minimum, the derivative is zero:

\[
2\sum_{i=1}^{m}(ax_i+b-y_i)x_i = 0
\]

Remove the 2:

\[
\sum_{i=1}^{m}(ax_i+b-y_i)x_i = 0
\]

Expand:

\[
a\sum_{i=1}^{m}x_i^2
+
b\sum_{i=1}^{m}x_i
-
\sum_{i=1}^{m}x_iy_i
=0
\]

Move the last term to the right side:

\[
a\sum_{i=1}^{m}x_i^2
+
b\sum_{i=1}^{m}x_i
=
\sum_{i=1}^{m}x_iy_i
\]

---

## Derivative with respect to \(b\)

\[
\frac{\partial f}{\partial b}
=
\sum_{i=1}^{m} 2r_i \frac{\partial r_i}{\partial b}
\]

Since:

\[
\frac{\partial r_i}{\partial b} = 1
\]

we get:

\[
\frac{\partial f}{\partial b}
=
2\sum_{i=1}^{m}(ax_i+b-y_i)
\]

At the minimum:

\[
2\sum_{i=1}^{m}(ax_i+b-y_i)=0
\]

Remove 2:

\[
\sum_{i=1}^{m}(ax_i+b-y_i)=0
\]

Expand:

\[
a\sum_{i=1}^{m}x_i
+
bm
-
\sum_{i=1}^{m}y_i
=0
\]

So:

\[
a\sum_{i=1}^{m}x_i
+
bm
=
\sum_{i=1}^{m}y_i
\]

---

## Final system

\[
\begin{bmatrix}
\sum x_i^2 & \sum x_i \\
\sum x_i & m
\end{bmatrix}
\begin{bmatrix}
a \\
b
\end{bmatrix}
=
\begin{bmatrix}
\sum x_iy_i \\
\sum y_i
\end{bmatrix}
\]

This is exactly the system we solve in Python.

# 12. Exercise Task 1(b): Function for Fitting a Line

Now we write a simple function.

Input:

- `x_values`
- `y_values`

Output:

- slope `a`
- intercept `b`

````python
def fit_line(x_values, y_values):
    x = np.array(x_values)
    y = np.array(y_values)

    m = len(x)

    sum_x2 = np.sum(x * x)
    sum_x = np.sum(x)
    sum_xy = np.sum(x * y)
    sum_y = np.sum(y)

    A = np.array([
        [sum_x2, sum_x],
        [sum_x,  m]
    ])

    right_side = np.array([sum_xy, sum_y])

    a, b = np.linalg.solve(A, right_side)

    return a, b
````

Let us test the function with a very small example.

Suppose our data is almost like:

\[
y = 2x + 1
\]

````python
x_test = np.array([0, 1, 2, 3])
y_test = np.array([1, 3, 5, 7])

a, b = fit_line(x_test, y_test)

print("a =", a)
print("b =", b)
````

The result should be close to:

\[
a = 2,\quad b = 1
\]

# 14. Exercise Task 1(d): Line Must Pass Through One Given Point

Now the line must pass through:

\[
(s,z)
\]

The line is still:

\[
g(x)=ax+b
\]

Passing through \((s,z)\) means:

\[
z = as + b
\]

So:

\[
b = z - as
\]

Now we do not need to search for both \(a\) and \(b\).  
We only search for \(a\).

Put \(b=z-as\) into the line:

\[
g(x_i)=ax_i+z-as
\]

\[
g(x_i)=a(x_i-s)+z
\]

The error is:

\[
a(x_i-s)+z-y_i
\]

So the new function is:

\[
F(a)=\sum_{i=1}^{m}(a(x_i-s)+z-y_i)^2
\]

Differentiate with respect to \(a\):

\[
\frac{dF}{da}
=
2\sum_{i=1}^{m}(a(x_i-s)+z-y_i)(x_i-s)
\]

At the minimum:

\[
\sum_{i=1}^{m}(a(x_i-s)+z-y_i)(x_i-s)=0
\]

Expand:

\[
a\sum_{i=1}^{m}(x_i-s)^2
+
\sum_{i=1}^{m}(z-y_i)(x_i-s)
=0
\]

So:

\[
a\sum_{i=1}^{m}(x_i-s)^2
=
\sum_{i=1}^{m}(y_i-z)(x_i-s)
\]

Therefore:

\[
a
=
\frac{\sum_{i=1}^{m}(x_i-s)(y_i-z)}
{\sum_{i=1}^{m}(x_i-s)^2}
\]

Then:

\[
b=z-as
\]

## Is there always a solution?

If

\[
\sum_{i=1}^{m}(x_i-s)^2 = 0
\]

then all \(x_i=s\).  
Then the slope \(a\) cannot be determined uniquely.

So:

- if at least one \(x_i \neq s\), we get one clear solution
- if all \(x_i=s\), the slope is not unique

````python
def fit_line_through_point(x_values, y_values, s, z):
    x = np.array(x_values)
    y = np.array(y_values)

    denominator = np.sum((x - s) ** 2)

    if denominator == 0:
        raise ValueError("No unique slope: all x-values are equal to s.")

    numerator = np.sum((x - s) * (y - z))

    a = numerator / denominator
    b = z - a * s

    return a, b
````

Let us choose a point \((s,z)\) and compare:

1. normal best-fit line
2. best-fit line forced through \((s,z)\)

````python
s = 0.5
z = 0.5

a_normal, b_normal = fit_line(x, y)
a_forced, b_forced = fit_line_through_point(x, y, s, z)

print("Normal line:")
print("a =", a_normal)
print("b =", b_normal)

print("\nLine through point:")
print("a =", a_forced)
print("b =", b_forced)
````

````python
x_line = np.linspace(min(x), max(x), 100)

y_normal = a_normal * x_line + b_normal
y_forced = a_forced * x_line + b_forced

plt.figure(figsize=(7, 5))
plt.scatter(x, y, label="data points")
plt.plot(x_line, y_normal, label="normal fitted line")
plt.plot(x_line, y_forced, label="line through chosen point")
plt.scatter([s], [z], s=100, marker="x", label="chosen point (s,z)")
plt.xlabel("x")
plt.ylabel("y")
plt.title("Normal fit vs. forced-through-point fit")
plt.legend()
plt.grid(True)
plt.show()
````

## Comparison

The normal line is completely free.

The forced line has one restriction: it must pass through the chosen point.  
Because of this, its total error may be larger than the normal line.

````python
def squared_error(x_values, y_values, a, b):
    x = np.array(x_values)
    y = np.array(y_values)

    prediction = a * x + b
    error = prediction - y

    return np.sum(error ** 2)

error_normal = squared_error(x, y, a_normal, b_normal)
error_forced = squared_error(x, y, a_forced, b_forced)

print("error normal line =", error_normal)
print("error forced line =", error_forced)
````

# 15. Wine Example: Linear Regression in Many Dimensions

The wine task has many input parameters.

For one wine, we may have:

1. fixed acidity
2. volatile acidity
3. citric acid
4. residual sugar
5. chlorides
6. free sulphur dioxide
7. total sulphur dioxide
8. density
9. pH
10. sulphates
11. alcohol

The output is:

\[
\text{quality}
\]

For many-dimensional linear regression, the model is:

\[
\hat{y} = a_1x_1 + a_2x_2 + \cdots + a_{11}x_{11} + b
\]

In matrix form:

\[
\hat{Y} = X a + b
\]

It is easier to add one extra column of ones to \(X\):

\[
X_{\text{new}} =
\begin{bmatrix}
x_{11} & x_{12} & \cdots & 1 \\
x_{21} & x_{22} & \cdots & 1 \\
\vdots & \vdots & \ddots & \vdots
\end{bmatrix}
\]

Then the unknown becomes:

\[
\theta =
\begin{bmatrix}
a_1 \\
a_2 \\
\vdots \\
b
\end{bmatrix}
\]

The normal equation is:

\[
X_{\text{new}}^T X_{\text{new}} \theta
=
X_{\text{new}}^T Y
\]

## Loading the wine data

The exercise mentions a file called `winedata.csv`.

Put `winedata.csv` in the same folder as this notebook.

The code below does this:

- if `winedata.csv` exists, it loads it
- if it does not exist, it creates a small artificial example so the notebook still runs

````python
import os
import pandas as pd

csv_file = "winedata.csv"

if os.path.exists(csv_file):
    data = pd.read_csv(csv_file)
    print("Loaded real wine data.")
else:
    print("winedata.csv not found.")
    print("Creating artificial example data so the notebook still runs.")

    np.random.seed(1)

    number_of_wines = 1507
    number_of_features = 11

    X_fake = np.random.random((number_of_wines, number_of_features))

    true_a = np.array([0.2, -0.4, 0.1, 0.05, -0.2, 0.01, -0.01, -0.3, 0.1, 0.25, 0.8])
    true_b = 5.0

    noise = 0.2 * np.random.randn(number_of_wines)
    Y_fake = X_fake @ true_a + true_b + noise

    column_names = [
        "fixed acidity",
        "volatile acidity",
        "citric acid",
        "residual sugar",
        "chlorides",
        "free sulphur dioxide",
        "total sulphur dioxide",
        "density",
        "pH",
        "sulphates",
        "alcohol",
        "quality"
    ]

    data = pd.DataFrame(
        np.column_stack([X_fake, Y_fake]),
        columns=column_names
    )

data.head()
````

## Separate input \(X\) and output \(Y\)

The first 11 columns are input features.  
The last column is the quality.

````python
X = data.iloc[:, 0:11].to_numpy()
Y = data.iloc[:, 11].to_numpy()

print("shape of X =", X.shape)
print("shape of Y =", Y.shape)
````

## Train-test split

Training data is used to find the parameters.

Test data is used to check if the model works on new data.

````python
np.random.seed(2)

m = len(Y)
indices = np.arange(m)
np.random.shuffle(indices)

train_size = int(0.8 * m)

train_indices = indices[:train_size]
test_indices = indices[train_size:]

X_train = X[train_indices]
Y_train = Y[train_indices]

X_test = X[test_indices]
Y_test = Y[test_indices]

print("training samples =", len(Y_train))
print("test samples =", len(Y_test))
````

## Fit the multidimensional linear regression by hand

We solve:

\[
X_{\text{new}}^T X_{\text{new}} \theta
=
X_{\text{new}}^T Y
\]

where the last parameter in \(\theta\) is \(b\).

````python
def fit_linear_regression_many_dimensions(X, Y):
    number_of_rows = X.shape[0]

    ones_column = np.ones((number_of_rows, 1))
    X_new = np.column_stack([X, ones_column])

    A = X_new.T @ X_new
    right_side = X_new.T @ Y

    # np.linalg.solve works when A is nicely invertible.
    # If there is a numerical problem, np.linalg.lstsq is safer.
    try:
        theta = np.linalg.solve(A, right_side)
    except np.linalg.LinAlgError:
        theta = np.linalg.lstsq(A, right_side, rcond=None)[0]

    a = theta[:-1]
    b = theta[-1]

    return a, b

a, b = fit_linear_regression_many_dimensions(X_train, Y_train)

print("a =", a)
print("b =", b)
````

## Prediction function

Now we predict:

\[
\hat{Y} = X a + b
\]

````python
def predict(X, a, b):
    return X @ a + b

Y_train_pred = predict(X_train, a, b)
Y_test_pred = predict(X_test, a, b)
````

## Check the error

We use simple error measures:

### Mean Absolute Error

\[
MAE = \frac{1}{m}\sum |y_i-\hat{y}_i|
\]

This means average absolute mistake.

### Root Mean Squared Error

\[
RMSE = \sqrt{\frac{1}{m}\sum (y_i-\hat{y}_i)^2}
\]

This gives larger punishment to big mistakes.

### \(R^2\) score

\(R^2\) tells how much variation is explained by the model.

- \(R^2 = 1\) is perfect
- \(R^2 = 0\) is not better than predicting the mean
- \(R^2 < 0\) is bad

````python
def mean_absolute_error(y_true, y_pred):
    return np.mean(np.abs(y_true - y_pred))

def root_mean_squared_error(y_true, y_pred):
    return np.sqrt(np.mean((y_true - y_pred) ** 2))

def r2_score_simple(y_true, y_pred):
    numerator = np.sum((y_true - y_pred) ** 2)
    denominator = np.sum((y_true - np.mean(y_true)) ** 2)
    return 1 - numerator / denominator

print("Training error")
print("MAE  =", mean_absolute_error(Y_train, Y_train_pred))
print("RMSE =", root_mean_squared_error(Y_train, Y_train_pred))
print("R2   =", r2_score_simple(Y_train, Y_train_pred))

print("\nTest error")
print("MAE  =", mean_absolute_error(Y_test, Y_test_pred))
print("RMSE =", root_mean_squared_error(Y_test, Y_test_pred))
print("R2   =", r2_score_simple(Y_test, Y_test_pred))
````

# 16. Small Human Summary

Linear regression is not magic.

It means:

1. Choose a model, for example:
   \[
   y = ax+b
   \]

2. Measure the error:
   \[
   ax_i+b-y_i
   \]

3. Square the errors and add them:
   \[
   \sum (ax_i+b-y_i)^2
   \]

4. Differentiate with respect to the unknowns.

5. Set derivatives equal to zero.

6. Solve the resulting linear system.

For many input variables, the idea is the same.  
Only the matrix becomes larger.
