## 4. Variables

A variable is like a small box where we store a value.

Example:

```python

```

Here:

- `name` is the variable
- `"Shakeel"` is the value stored inside the variable

## 5. Rules for Variable Names

Good variable names make code easy to read.

### Allowed

```python
age = 25
student_name = "Ali"
total_marks = 90
```

### Not allowed

```python
2name = "Ali"
student-name = "Ali"
total marks = 90
```

Important rules:

- Variable names cannot start with a number.
- Use underscore `_` instead of spaces.
- Variable names are case-sensitive.

````python
name = "Shakeel"
Name = "Python"

print(name)
print(Name)
````

`name` and `Name` are different variables because Python is case-sensitive.

# Part 2: Data Types

A data type tells Python what kind of value we are using.

Common data types:

| Data Type | Example | Meaning |
|---|---|---|
| `int` | `10` | Whole number |
| `float` | `10.5` | Decimal number |
| `str` | `"Python"` | Text |
| `bool` | `True` | True or False |
| `list` | `[1, 2, 3]` | Ordered collection |
| `tuple` | `(1, 2, 3)` | Ordered collection that cannot be changed |
| `set` | `{1, 2, 3}` | Unique values |
| `dict` | `{"name": "Ali"}` | Key-value data |
| `NoneType` | `None` | No value |

## 7. Integer Data Type

An integer is a whole number.

Examples:

- `5`
- `100`
- `-20`

````python
age = 25
year = 2026
temperature = -5

print(age)
print(year)
print(temperature)

print(type(age))
````

`type()` tells us the data type of a variable.

## 8. Float Data Type

A float is a decimal number.

Examples:

- `3.14`
- `10.5`
- `-2.75`

````python
height = 1.75
price = 9.99
pi = 3.14159

print(height)
print(price)
print(pi)

print(type(height))
````

## 9. String Data Type

A string is text.

Strings are written inside quotes.

You can use:

- single quotes: `'Python'`
- double quotes: `"Python"`

````python
language = "Python"
message = 'Python is easy to start'

print(language)
print(message)

print(type(language))
````

### String indexing

Each character in a string has a position.

Python counting starts from `0`.

For example:

```python
P y t h o n
0 1 2 3 4 5
```

````python
word = "Python"

print(word[0])
print(word[1])
print(word[2])
````

### String slicing

Slicing means taking a part of a string.

````python
word = "Python Programming"

print(word[0:6])
print(word[7:18])
````

## 10. Boolean Data Type

A boolean has only two possible values:

- `True`
- `False`

Booleans are often used in conditions.

````python
is_student = True
is_raining = False

print(is_student)
print(is_raining)

print(type(is_student))
````

## 11. List Data Type

A list stores many values in one variable.

Lists are written using square brackets `[]`.

````python
marks = [80, 75, 90, 85]

print(marks)
print(type(marks))
````

Lists are ordered.

That means each value has an index position.

````python
marks = [80, 75, 90, 85]

print(marks[0])
print(marks[1])
print(marks[2])
````

Lists can be changed.

````python
fruits = ["apple", "banana", "mango"]

print("Before:", fruits)

fruits.append("orange")

print("After:", fruits)
````

## 12. Tuple Data Type

A tuple is similar to a list.

But a tuple cannot be changed after creation.

Tuples are written using round brackets `()`.

````python
coordinates = (10, 20)

print(coordinates)
print(type(coordinates))
print(coordinates[0])
````

Use tuples when the values should stay fixed.

Example:

- coordinates
- RGB color values
- fixed dimensions

## 13. Set Data Type

A set stores unique values.

Sets are written using curly brackets `{}`.

Duplicate values are automatically removed.

````python
numbers = {1, 2, 3, 3, 4, 4, 5}

print(numbers)
print(type(numbers))
````

Sets are useful when you only care about unique items.

````python
students = {"Ali", "John", "Ali", "Sara"}

print(students)
````

## 14. Dictionary Data Type

A dictionary stores data in key-value pairs.

Example:

```python
student = {
    "name": "Ali",
    "age": 22
}
```

Here:

- `"name"` is a key
- `"Ali"` is the value

````python
student = {
    "name": "Shakeel",
    "age": 25,
    "city": "Dresden"
}

print(student)
print(type(student))
````

We can access dictionary values using keys.

````python
student = {
    "name": "Shakeel",
    "age": 25,
    "city": "Dresden"
}

print(student["name"])
print(student["age"])
print(student["city"])
````

## 15. None Data Type

`None` means no value.

It is useful when we want to create an empty variable first.

````python
result = None

print(result)
print(type(result))
````

Example:

Maybe we will calculate a result later, but for now we do not have the value.

````python
final_answer = None

# Later in the program
final_answer = 100

print(final_answer)
````

# Part 3: Operators

Operators are symbols used to perform operations.

Example:

```python
10 + 5
```

Here, `+` is an operator.

## 16. Arithmetic Operators

Arithmetic operators are used for mathematical calculations.

| Operator | Meaning | Example |
|---|---|---|
| `+` | Addition | `10 + 5` |
| `-` | Subtraction | `10 - 5` |
| `*` | Multiplication | `10 * 5` |
| `/` | Division | `10 / 5` |
| `//` | Floor division | `10 // 3` |
| `%` | Remainder | `10 % 3` |
| `**` | Power | `2 ** 3` |

````python
a = 10
b = 3

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor division:", a // b)
print("Remainder:", a % b)
print("Power:", a ** b)
````

### Important difference: `/` and `//`

`/` gives normal division.

`//` gives floor division, meaning it removes the decimal part.

````python
print(10 / 3)
print(10 // 3)
````

### Remainder operator `%`

The `%` operator gives the remainder after division.

````python
print(10 % 3)
print(12 % 5)
print(20 % 2)
````

A common use of `%` is checking even or odd numbers.

````python
number = 10

if number % 2 == 0:
    print("Even number")
else:
    print("Odd number")
````

## 17. Assignment Operators

Assignment operators are used to store or update values.

| Operator | Example | Same as |
|---|---|---|
| `=` | `x = 5` | Store value |
| `+=` | `x += 2` | `x = x + 2` |
| `-=` | `x -= 2` | `x = x - 2` |
| `*=` | `x *= 2` | `x = x * 2` |
| `/=` | `x /= 2` | `x = x / 2` |

````python
x = 10

print("Start:", x)

x += 5
print("After x += 5:", x)

x -= 3
print("After x -= 3:", x)

x *= 2
print("After x *= 2:", x)

x /= 4
print("After x /= 4:", x)
````

## 18. Comparison Operators

Comparison operators compare two values.

They return either `True` or `False`.

| Operator | Meaning |
|---|---|
| `==` | Equal to |
| `!=` | Not equal to |
| `>` | Greater than |
| `<` | Less than |
| `>=` | Greater than or equal to |
| `<=` | Less than or equal to |

````python
a = 10
b = 5

print(a == b)
print(a != b)
print(a > b)
print(a < b)
print(a >= b)
print(a <= b)
````

Comparison operators are very useful in decision making.

````python
marks = 75

print(marks >= 50)
````

## 19. Logical Operators

Logical operators combine conditions.

| Operator | Meaning |
|---|---|
| `and` | True only if both conditions are True |
| `or` | True if at least one condition is True |
| `not` | Reverses True and False |

````python
age = 22
has_id = True

print(age >= 18 and has_id == True)
````

### `and` example

Both conditions must be true.

````python
temperature = 25
is_sunny = True

if temperature > 20 and is_sunny:
    print("Good weather for a walk")
else:
    print("Maybe stay inside")
````

### `or` example

At least one condition must be true.

````python
has_cash = False
has_card = True

if has_cash or has_card:
    print("You can buy the item")
else:
    print("You cannot buy the item")
````

### `not` example

`not` reverses the boolean value.

````python
is_raining = False

print(not is_raining)
````

## 20. Membership Operators

Membership operators check whether something exists inside a collection.

| Operator | Meaning |
|---|---|
| `in` | Exists inside |
| `not in` | Does not exist inside |

````python
fruits = ["apple", "banana", "mango"]

print("apple" in fruits)
print("orange" in fruits)
print("orange" not in fruits)
````

This is useful when checking lists, strings, sets, or dictionaries.

````python
sentence = "Python is a powerful language"

print("Python" in sentence)
print("Java" in sentence)
````

## 21. Identity Operators

Identity operators check whether two variables refer to the same object.

| Operator | Meaning |
|---|---|
| `is` | Same object |
| `is not` | Not the same object |

````python
a = [1, 2, 3]
b = a
c = [1, 2, 3]

print(a is b)
print(a is c)
print(a == c)
````

Important:

- `==` checks if values are equal.
- `is` checks if both variables point to the same object in memory.

## Example 5: Engineering Style Example

Suppose a force is applied on an area.

Stress is:

\[
\text{stress} = \frac{\text{force}}{\text{area}}
\]

This is common in mechanics.

````python
force = 1000      # Newton
area = 0.005      # square meter

stress = force / area

print("Stress =", stress, "Pa")
````
